### Eserrcizio 1

Data la lista dei voti di una verifica, stampa il numero di voti, la somma, la media con due cifre
decimali, il voto minimo, il voto massimo e quanti voti sono sufficienti, cioè maggiori o uguali a
sei. Usa len , sum , min , max e un ciclo for con un contatore per le sufficienze. Sezione
del notebook: es01_statistiche

```python
voti = [7, 5, 8, 6, 9, 4, 7, 10]

n = 0 #numero di voti
v = 0 #somma dei voti
for voto in voti:
    n += 1
    v = v+voto

m = v/n
print(n)
print(v)
print(m)
```
