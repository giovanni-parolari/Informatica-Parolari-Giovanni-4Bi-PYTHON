Scrivere nella cella dell'esercizio 1 il programma che legge in ingresso la durata di una sessione di laboratorio espressa in secondi e la stampa nel formato ore, minuti e secondi, usando soltanto la divisione intera e il resto. Il programma stampa anche la stessa durata espressa in MODULO M2 — ESERCIZI SUI FONDAMENTI DEL LINGUAGGIO PYTHON LABORATORIO DI INFORMATICA — 4BI 2 ore decimali con due cifre dopo la virgola, e segnala con un messaggio il caso in cui la durata superi le quattro ore.

```python
d = 4250  # Dato iniziale in secondi

# Conversione di secondi in ore, minuti e secondi
ore = d // 3600  # Prende solo la parte intera delle ore (1)
minuti = (d % 3600) // 60  # Prende il resto delle ore e calcola i minuti interi (10)
secondi = d % 60  # Prende il resto finale dei secondi (50)

print(f"Ore: {ore}")
print(f"Minuti: {minuti}")
print(f"Secondi: {secondi}")
print(f"ore,minuti, secondi   
      {ore}:{minuti}:{secondi}")
```
