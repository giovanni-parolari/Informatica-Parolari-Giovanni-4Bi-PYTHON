# Esercizio 1


Scrivere nella cella dell'esercizio 1 il programma che legge in ingresso la durata di una sessione di laboratorio espressa in secondi e la stampa nel formato ore, minuti e secondi, usando soltanto la divisione intera e il resto. Il programma stampa anche la stessa durata espressa in MODULO M2 — ESERCIZI SUI FONDAMENTI DEL LINGUAGGIO PYTHON LABORATORIO DI INFORMATICA — 4BI 2 ore decimali con due cifre dopo la virgola, e segnala con un messaggio il caso in cui la durata superi le quattro ore.

```python
d = input("Inserisci un numero")  # Dato iniziale in secondi

# Conversione di secondi in ore, minuti e secondi
ore = int(d) // 3600  # Prende solo la parte intera delle ore (1)
minuti = (int(d) % 3600) // 60  # Prende il resto delle ore e calcola i minuti interi (10)
secondi = int(d) % 60  # Prende il resto finale dei secondi (50)

print(f"ore, minuti, secondi")
print(f"{ore}:{minuti}:{secondi}")
```

secondo questa regola: se le ore di assenza superano 50 il giudizio è non valutabile , indipendentemente dalla media; altrimenti il giudizio è ottimo da 8 in su, sufficiente da 6 in
su, insufficiente sotto il 6. La media va stampata con due decimali dentro una f-string,
insieme al nome dello studente.

```python
import sys

studente = input("Inserisci il nome")
ore = input("Inserisci le ore di assenza")

if int(ore) > 50:
    print("L'alunno non può essere scrutianto perche ha superato il limite massimo di ore")
    sys.exit()
    

v1 = input("Inserisci primo voto")
v2 = input("Inserisci secondo voto")
v3 = input("Inserisci terzo voto")

media = float(v1+v2+v3)/3
if media >8:
    giudizio = ("Ottimo")
elif media > 6:
    giudizio = ("Sufficiente")
elif media <6:    
    giudizio = ("Insufficente")
    
print(f"Media {media}, giudizio {studente}: {giudizio}")
```

# Esercizio 3

Scrivere nella cella dell'esercizio 3 il programma che legge in ingresso un valore iniziale, un
valore finale e un passo, tutti interi, e stampa la tavola di conversione da chilometri a miglia e a
metri per tutti i valori dell'intervallo, estremi compresi. Un chilometro corrisponde a 0,621371
miglia. La tabella ha un'intestazione e una riga di trattini, le colonne sono allineate a destra e i
valori in miglia hanno tre decimali. Se il passo non è positivo il programma stampa un
messaggio e non produce la tabella.

```python
v1 = int(input("Inserisci valore iniziale: "))
v2 = int(input("Inserisci valore finale: "))
v3 = int(input("Inserisci un passo: "))

if v3 <= 0:
    print("Il passo deve essere positivo.")
else:
    print(f"{'Chilometri':>12} {'Miglia':>10} {'Metri':>10}")
    print("-" * 34)
    for km in range(v1, v2 + 1, v3):
        miglia = km * 0.621371
        metri = km * 1000
        print(f"{km:12d} {miglia:10.3f} {metri:10d}")
```
