# Esercizio 7

Scrivere nella cella dell'esercizio 7 il programma che legge in ingresso un intero non negativo e ne calcola, con un ciclo while che usa soltanto il resto e la divisione intera per dieci, la somma delle cifre, il numero di cifre e la cifra maggiore. Il numero zero va gestito correttamente: ha una cifra, la somma è zero e la cifra maggiore è zero. Non è ammesso trasformare il
numero in stringa.

```python
# Legge un intero non negativo in ingresso
numero = int(input("Inserisci un numero intero non negativo: "))

# Inizializzazione delle variabili
somma_cifre = 0
numero_cifre = 0
cifra_massima = 0

# Gestione del caso in cui il numero sia zero
if numero == 0:
    somma_cifre = 0
    numero_cifre = 1
    cifra_massima = 0
else:
    # Ciclo while per calcolare somma delle cifre, numero di cifre e cifra maggiore
    while numero > 0:
        cifra = numero % 10  # Estrae l'ultima cifra
        somma_cifre += cifra  # Aggiunge la cifra alla somma
        numero_cifre += 1  # Incrementa il conteggio delle cifre
        if cifra > cifra_massima:  # Aggiorna la cifra massima se necessario
            cifra_massima = cifra
        numero //= 10  # Rimuove l'ultima cifra

# Stampa dei risultati
print("Somma delle cifre:", somma_cifre)
print("Numero di cifre:", numero_cifre)
print("Cifra maggiore:", cifra_massima)
```

# Esercizio 8 

Scrivere nella cella dell'esercizio 8 il programma che legge in ingresso un intero non negativo, ne costruisce la rappresentazione binaria come stringa con il metodo dei resti successivi, e la confronta con il risultato della funzione predefinita bin , stampando verifica superata oppure verifica fallita . Il programma legge poi una seconda stringa composta di soli zeri e uni e ne calcola il valore decimale scorrendone i caratteri da sinistra a destra, senza usare la conversione predefinita. Se la seconda stringa contiene caratteri diversi da 0 e 1 , il programma lo segnala e non calcola nulla.

```python
# Funzione per calcolare la rappresentazione binaria con il metodo dei resti# successivi
def calcola_binario(n):
    if n == 0:
        return "0"
    binario = ""
    while n > 0:
        resto = n % 2
        binario = str(resto) + binario
        n = n // 2
    return binario

# Programma principale
# Lettura di un intero non negativo
numero = int(input("Inserisci un intero non negativo: "))
if numero < 0:
    print("Errore: il numero deve essere non negativo.")
else:
    # Calcolo della rappresentazione binaria
    binario_calcolato = calcola_binario(numero)
    binario_predefinito = bin(numero)[2:]  # Rimuove il prefisso '0b'

    # Confronto tra le due rappresentazioni
    if binario_calcolato == binario_predefinito:
        print("Verifica superata")
    else:
        print("Verifica fallita")

# Lettura di una stringa composta di soli zeri e uni
stringa_binaria = input("Inserisci una stringa di soli 0 e 1: ")

# Controllo della validità della stringa
if not all(c in "01" for c in stringa_binaria):
    print("Errore: la stringa contiene caratteri diversi da 0 e 1.")
else:
    # Calcolo del valore decimale scorrendo i caratteri
    valore_decimale = 0
    for carattere in stringa_binaria:
        valore_decimale = valore_decimale * 2 + int(carattere)
    print(f"Il valore decimale della stringa è: {valore_decimale}")
```

# Esercizio 10

Scrivere nella cella dell'esercizio 10 il programma che legge in ingresso una sequenza di voti in
virgola mobile, uno per riga, e termina quando l'utente digita 0 . Il programma calcola e stampa il numero di voti inseriti, il massimo, il minimo, la media con due decimali e quanti voti sono risultati sufficienti, cioè maggiori o uguali a 6. I voti fuori dall'intervallo da 1 a 10 vanno
segnalati e ignorati senza interrompere la lettura. Se non viene inserito alcun voto valido il
programma lo dichiara e non calcola le statistiche.

```python
# Programma per leggere una sequenza di voti e calcolare statistiche

voti = []
sufficienti = 0

while True:
    try:
        voto = float(input("Inserisci un voto (0 per terminare): "))
        if voto == 0:
            break
        if voto < 1 or voto > 10:
            print("Voto fuori dall'intervallo [1, 10]. Ignorato.")
            continue
        voti.append(voto)
        if voto >= 6:
            sufficienti += 1
    except ValueError:
        print("Input non valido. Inserisci un numero.")

if voti:
    massimo = max(voti)
    minimo = min(voti)
    media = sum(voti) / len(voti)
    print(f"Numero di voti validi: {len(voti)}")
    print(f"Massimo: {massimo}")
    print(f"Minimo: {minimo}")
    print(f"Media: {media:.2f}")
    print(f"Voti sufficienti: {sufficienti}")
else:
    print("Nessun voto valido inserito.")
```

# Esercizio 11

Scrivere nella cella dell'esercizio 11 il programma che legge in ingresso una frase, ne costruisce la versione normalizzata tenendo soltanto le lettere e portandole in minuscolo, e verifica se è palindroma con il confronto fra due indici che si avvicinano dagli estremi verso il centro,
interrompendo il ciclo al primo carattere diverso. Lo slicing con passo negativo va usato solo
alla fine, per stampare una riga di controllo che confronta il risultato ottenuto con quello dato
dal confronto ripulita >= ripulita[>:-1] . Il programma stampa anche quanti confronti
sono stati eseguiti prima di decidere.

```python
# Programma per verificare se una frase è palindroma

# Lettura della frase
frase = input("Inserisci una frase: ")

# Normalizzazione: tenere solo lettere e convertirle in minuscolo
ripulita = ''.join(c.lower() for c in frase if c.isalpha())

# Verifica se è palindroma
inizio = 0
fine = len(ripulita) - 1
confronti = 0
palindroma = True

while inizio < fine:
    confronti += 1
    if ripulita[inizio] != ripulita[fine]:
        palindroma = False
        break
    inizio += 1
    fine -= 1

# Stampa del risultato
if palindroma:
    print("La frase è palindroma.")
else:
    print("La frase non è palindroma.")

# Stampa del numero di confronti
print(f"Confronti eseguiti: {confronti}")

# Riga di controllo con slicing
print(f"Controllo con slicing: {ripulita == ripulita[::-1]}")
```

# Esercizio 12

Scrivere nella cella dell'esercizio 12 il programma che legge in ingresso un testo e una sotto‐
stringa da cercare, e conta quante volte la sottostringa compare nel testo usando find ripetutamente dentro un ciclo, senza chiamare count . Il programma stampa il numero di occorrenze e la posizione di ciascuna, contando da zero. Alla fine stampa la stessa cosa calcolata
con count , come verifica del proprio risultato, e stampa il testo con tutte le occorrenze sostituite da un asterisco per ciascun carattere della sottostringa, ottenuto ricostruendo la stringa
con lo slicing e la concatenazione, senza replace.

```python
# Programma per contare le occorrenze di una sottostringa e sostituirle con asterischi

# Lettura del testo e della sottostringa
testo = input("Inserisci il testo: ")
sottostringa = input("Inserisci la sottostringa da cercare: ")

# Contare le occorrenze usando find ripetutamente
posizione = 0
occorrenze = 0
posizioni = []

while True:
    posizione = testo.find(sottostringa, posizione)
    if posizione == -1:
        break
    posizioni.append(posizione)
    occorrenze += 1
    posizione += 1  # Spostarsi oltre la posizione corrente per continuare la ricerca

# Stampa delle occorrenze e delle posizioni
print(f"Numero di occorrenze trovate: {occorrenze}")
print(f"Posizioni: {posizioni}")

# Verifica con count
occorrenze_count = testo.count(sottostringa)
print(f"Verifica con count: {occorrenze_count}")

# Sostituire le occorrenze con asterischi
lunghezza_sottostringa = len(sottostringa)
testo_modificato = ""
inizio = 0

for posizione in posizioni:
    testo_modificato += testo[inizio:posizione] + '*' * lunghezza_sottostringa
    inizio = posizione + lunghezza_sottostringa

testo_modificato += testo[inizio:]  # Aggiungere il resto del testo

# Stampa del testo modificato
print("Testo modificato:")
print(testo_modificato)
```

# Esercizio 13

Il metodo seguente è stato scritto in Java al terzo anno. Riceve una stringa di voti espressi con
una sola cifra, ignora i caratteri che non sono cifre, e restituisce la percentuale di voti
sufficienti sul totale dei voti validi.

```python
public class Voti {
    public static double calcolaPercentualeSufficienti(String voti) {
        int validi = 0;
        int sufficienti = 0;

        for (int i = 0; i < voti.length(); i++) {
            char c = voti.charAt(i);
            if (Character.isDigit(c)) {
                validi++;
                int voto = Character.getNumericValue(c);
                if (voto >= 6) {
                    sufficienti++;
                }
            }
        }

        if (validi == 0) {
            return 0.0; // Nessun voto valido
        }

        return (sufficienti * 100.0) / validi;
    }

    public static void main(String[] args) {
        String voti = "8a5b6c7d4e9";
        double percentuale = calcolaPercentualeSufficienti(voti);
        System.out.println("Percentuale di voti sufficienti: " + percentuale + "%");
    }
}
```

In Java l'ordinamento di un array di numeri si è scritto con lo scambio fra elementi, tipicamente
con l'algoritmo a bolle o per selezione. Senza le liste quella tecnica non è disponibile, ma per
un insieme di valori che appartengono a un intervallo piccolo e noto esiste un'alternativa: contare quante volte compare ciascun valore possibile e ricostruire il risultato in ordine. È l'idea
dell'ordinamento per conteggio.
Scrivere nella cella dell'esercizio 14 il programma che legge in ingresso una stringa di sole
cifre e ne stampa le cifre in ordine crescente, ottenendo il risultato con un ciclo for sulle die‐
ci cifre possibili e con il conteggio delle occorrenze di ciascuna. Il programma stampa anche la
distribuzione delle cifre presenti, una riga per cifra, con un istogramma fatto di asterischi. Se
l'ingresso contiene caratteri che non sono cifre, il programma lo segnala e si ferma.
Nel commento finale della cella vanno scritte tre righe che confrontano questo procedimento
con l'ordinamento a bolle scritto in Java: quale dei due esegue meno confronti, perché questo
qui non è applicabile a un insieme

```python
# Programma per ordinare una stringa di cifre e stampare un istogramma

# Lettura della stringa di ingresso
stringa = input("Inserisci una stringa di sole cifre: ")

# Verifica che la stringa contenga solo cifre
if not stringa.isdigit():
    print("Errore: la stringa contiene caratteri non numerici.")
else:
    # Conteggio delle occorrenze di ciascuna cifra
    conteggio = [0] * 10  # Array per contare le cifre da 0 a 9
    for c in stringa:
        conteggio[int(c)] += 1

    # Ricostruzione della stringa ordinata
    ordinata = ""
    for i in range(10):  # Itera sulle cifre da 0 a 9
        ordinata += str(i) * conteggio[i]

    # Stampa della stringa ordinata
    print("Stringa ordinata:", ordinata)

    # Stampa della distribuzione delle cifre con istogramma
    print("Distribuzione delle cifre:")
    for i in range(10):
        if conteggio[i] > 0:
            print(f"{i}: {'*' * conteggio[i]}")

# Commento finale:
# 1. L'ordinamento per conteggio esegue meno confronti rispetto all'algoritmo a bolle,
#    poiché non confronta direttamente i valori tra loro, ma conta le occorrenze.
# 2. Questo metodo è applicabile solo quando i valori appartengono a un intervallo noto e limitato.
# 3. L'algoritmo a bolle è più generale, ma meno efficiente per insiemi grandi o con molti confronti.
```

calcola sommando le prime dodici cifre dopo averle moltiplicate alternativamente per uno e
per tre, a partire dal peso uno per la prima cifra, e prendendo poi la differenza fra dieci e il
resto della divisione per dieci di quella somma; se la differenza vale dieci, la cifra di controllo è
zero.
Scrivere nella cella dell'esercizio 15 il programma che legge in ingresso un codice, ne verifica
la lunghezza e la presenza di sole cifre, calcola la cifra di controllo a partire dalle prime dodici
e la confronta con la tredicesima, dichiarando il codice valido o non valido. In caso di codice
non valido va stampata anche la cifra di controllo corretta. Il programma stampa in ogni caso
la somma pesata calcolata, perché serve a correggere l'esercizio.

```python
# Programma per verificare la validità di un codice a barre EAN-13

# Lettura del codice
codice = input("Inserisci il codice EAN-13: ")

# Verifica della lunghezza e della presenza di sole cifre
if len(codice) != 13 or not codice.isdigit():
    print("Errore: il codice deve essere composto da 13 cifre.")
else:
    # Calcolo della somma pesata
    somma_pesata = 0
    for i in range(12):  # Considera solo le prime 12 cifre
        cifra = int(codice[i])
        if i % 2 == 0:  # Peso 1 per le cifre in posizione dispari (indice pari)
            somma_pesata += cifra
        else:  # Peso 3 per le cifre in posizione pari (indice dispari)
            somma_pesata += cifra * 3

    # Calcolo della cifra di controllo
    resto = somma_pesata % 10
    cifra_di_controllo_calcolata = (10 - resto) if resto != 0 else 0

    # Confronto con la tredicesima cifra
    cifra_di_controllo_originale = int(codice[12])
    print(f"Somma pesata calcolata: {somma_pesata}")
    if cifra_di_controllo_calcolata == cifra_di_controllo_originale:
        print("Il codice EAN-13 è valido.")
    else:
        print("Il codice EAN-13 non è valido.")
        print(f"Cifra di controllo corretta: {cifra_di_controllo_calcolata}")
```

# Esercizio 16

Scrivere nella cella dell'esercizio 16 il programma che legge in ingresso un testo e uno sposta‐
mento intero, e stampa il testo cifrato con il metodo di Cesare: ogni lettera viene sostituita
dalla lettera che si trova nello spostamento indicato più avanti nell'alfabeto, ricominciando
dall'inizio quando si supera la zeta. Le lettere maiuscole vanno portate in minuscolo prima
della cifratura; gli spazi e i segni di punteggiatura restano invariati.

```python
# Programma per cifrare un testo con il metodo di Cesare

# Lettura del testo e dello spostamento
testo = input("Inserisci il testo da cifrare: ")
spostamento = int(input("Inserisci lo spostamento: "))

# Funzione per cifrare una lettera
def cifra_lettera(lettera, spostamento):
    if lettera.isalpha():  # Controlla se è una lettera
        lettera = lettera.lower()  # Porta la lettera in minuscolo
        posizione = ord(lettera) - ord('a')  # Calcola la posizione nell'alfabeto (0-25)
        nuova_posizione = (posizione + spostamento) % 26  # Applica lo spostamento
        return chr(nuova_posizione + ord('a'))  # Ritorna la nuova lettera
    return lettera  # Restituisce il carattere invariato se non è una lettera

# Cifratura del testo
testo_cifrato = ''.join(cifra_lettera(c, spostamento) for c in testo)

# Stampa del risultato
print("Testo cifrato:", testo_cifrato)
```

In Java la ricerca binaria si è scritta su un array ordinato, con i due estremi che si avvicinano e
l'elemento centrale confrontato con il valore cercato. L'algoritmo si trasferisce in Python senza
alcuna modifica logica: cambia soltanto il modo di raggiungere la voce di posizione centrale,
perché l'array non è disponibile e l'elenco è memorizzato in una sola stringa.

```python
# Programma per eseguire la ricerca binaria su una stringa ordinata

def ricerca_binaria(stringa, valore):
    inizio = 0
    fine = len(stringa) - 1

    while inizio <= fine:
        # Calcolo della posizione centrale
        centro = (inizio + fine) // 2
        carattere_centrale = stringa[centro]

        # Confronto con il valore cercato
        if carattere_centrale == valore:
            return centro  # Valore trovato, restituisce la posizione
        elif carattere_centrale < valore:
            inizio = centro + 1  # Cerca nella metà destra
        else:
            fine = centro - 1  # Cerca nella metà sinistra

    return -1  # Valore non trovato

# Esempio di utilizzo
stringa_ordinata = "abcdefgijklmnop"
valore_cercato = input("Inserisci il carattere da cercare: ")

posizione = ricerca_binaria(stringa_ordinata, valore_cercato)

if posizione != -1:
    print(f"Il carattere '{valore_cercato}' si trova alla posizione {posizione}.")
else:
    print(f"Il carattere '{valore_cercato}' non è presente nella stringa.")
```

# Esercizio 18

Il registro delle presenze di una settimana è memorizzato in una sola stringa: le righe sono
separate dalla barra verticale e i campi di ogni riga dal punto e virgola, nell'ordine data, classe,
ore di lezione e ore di assenza complessive.

```python
# Programma per elaborare il registro delle presenze

# Stringa di esempio del registro
registro = "2023-10-01;1A;6;2|2023-10-02;2B;5;1|2023-10-03;3C;4;0|2023-10-04;1A;6;3|2023-10-05;2B;5;2"

# Suddividere il registro in righe
righe = registro.split('|')

# Elaborazione delle righe
for riga in righe:
    campi = riga.split(';')  # Suddividere i campi della riga
    data = campi[0]
    classe = campi[1]
    ore_lezione = int(campi[2])
    ore_assenza = int(campi[3])

    # Stampa dei dati elaborati
    print(f"Data: {data}, Classe: {classe}, Ore di lezione: {ore_lezione}, Ore di assenza: {ore_assenza}")
```
